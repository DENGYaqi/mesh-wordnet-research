```python
#-------------Cause we dont need the concept doesn exist in WN
#-------------So i use concept MeSH to check if it exist in WN or not.
#--------------using form origin and flexion.

# return list synset found
def find_word_in_WN(word):
    return wn.lemmas(word,lang='eng')

# return word of synset found in WN
def lemmatizer_and_stemmer_word(word):
    porter_stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()
    word_lemma = lemmatizer.lemmatize(word)
    return porter_stemmer.stem(word_lemma)

# return list of name concept MeSH not exist in WN(means no need)
def find_concept_MeSH_doesn_exist_in_WN(dataConceptMeSH, head_name):
    concept_not_exist = []
    for i in range(dataConceptMeSH.shape[0]):
        word_origin = dataConceptMeSH[head_name][i]
        synset_found_by_origin = find_word_in_WN(lemmatizer_and_stemmer_word(word_origin)) #example : Neoplasms, narrower, wn cant found by form flexion.
        synset_found_by_flexion = find_word_in_WN(word_origin) #example : Heterocyclic, wn cant found by form original.
        if len(synset_found_by_origin) == 0 and len(synset_found_by_flexion) == 0 :
            concept_not_exist.append(word_origin)
    return concept_not_exist
```

```python
#Extrat a tree number has the same length
def len_of_all_lever_tree_number():
    all_len_tree = []
    for i in range(1, dataConceptMeSH.shape[0]):
        getData = dataConceptMeSH['tree'][i].replace(' ', '')
        strToList = re.findall(r'[[](.*?)[]]', getData)[0].strip('[')
        target_list = [x for x in strToList.split(',')]
        #print(target_list)
        for j in range(len(target_list)):
            target_list[j] = target_list[j][1:-1]
            #print(target_list[j])
            #if(len(target_list[j] == 4)):
            #    print(target_list[j])
            all_len_tree.append(len(target_list[j]))
    return all_len_tree

def find_lever_of_tree():
    list_long_tree_number = len_of_all_lever_tree_number()
    list_remove_same_len = list(set(list_long_tree_number))
    list_remove_same_len.remove(4)#i dont know why len(tree_number) has 4
    lever_of_tree = sorted(list_remove_same_len, key=int, reverse=True)
    #inverse of Tree lever
    return lever_of_tree
```

```python
def find_a_lever_of_tree(lever_tree):
    lowest_list = []
    for i in range(1, dataConceptMeSH.shape[0]):
        getData = dataConceptMeSH['tree'][i].replace(' ', '')
        strToList = re.findall(r'[[](.*?)[]]', getData)[0].strip('[')
        target_list = [x for x in strToList.split(',')]
        for j in range(len(target_list)):
            des_lewest_every_lever = []
            target_list[j] = target_list[j][1:-1]
            if(len(target_list[j]) == lever_tree):
                lowest_list.append(target_list[j])
    return lowest_list
```

```python
def find_all_lever_of_tree():
    lever_tree = find_lever_of_tree()
    all_lever_class_by_lever_tree = []
    for i in lever_tree:
        all_lever_class_by_lever_tree.append(find_a_lever_of_tree(i))
    return all_lever_class_by_lever_tree
```
